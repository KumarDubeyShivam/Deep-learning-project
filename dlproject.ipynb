```python
import os
import requests
import argparse
import sys # Needed for detecting the interactive environment

# Define constants for the Google Drive IDs
IDS = {
    'train_sharp': '1YLksKtMhd2mWyVSkvhDaDLWSc1qYNCz-',
    'train_blur': '1Be2cgzuuXibcqAuJekDgvHq4MLYkCgR8',
    'train_blur_comp': '1hi6348BB9QQFqVx2PY7pKn32HQM89CJ1',
    'train_sharp_bicubic': '1a4PrjqT-hShvY9IyJm3sPF0ZaXyrCozR',
    'train_blur_bicubic': '10u8gthv2Q95RMCb1LeCN8N4ozB8TVjMt',
    'val_sharp': '1MGeObVQ1-Z29f-myDP7-8c3u0_xECKXq',
    'val_blur': '1N8z2yD0GDWmh6U4d4EADERtcUgDzGrHx',
    'val_blur_comp': '13d1uzqLdbsQzeZkWgdF5QVHqDSjfE4zZ',
    'val_sharp_bicubic': '1sChhtzN9Css10gX7Xsmc2JaC-2Pzco6a',
    'val_blur_bicubic': '1i3NAb7EmF4fCYadGaHK54-Zgx9lIC2Gp',
    'test_blur': '1dr0--ZBKqr4P1M8lek6JKD1Vd6bhhrZT',
    'test_blur_comp': '1OctyKR3ER_YWrZxKxQsZzLis3BvLSOFO',
    'test_sharp_bicubic': '1y0Jle6xB41TdRK_QMJ_E8W_iBMxwq_Rh',
    'test_blur_bicubic': '14YszfzUAeAfwP0ZA2FRzAiVxxZLg7-tY',
}

def download_file_from_server(server, subset, destination):
    """
    Downloads a specific subset of the REDS dataset either from Google Drive
    or the SNU CVLab server, handling large file warnings for Google Drive.
    """

    # Helper to find the Google Drive confirmation token for large files
    def get_confirm_token(response):
        for key, value in response.cookies.items():
            if key.startswith('download_warning'):
                return value
        return None

    # Helper to save the response content in chunks
    def save_response_content(response, destination):
        CHUNK_SIZE = 32768

        # Check for non-200 status codes (e.g., file not found, permission error)
        if response.status_code != 200:
            print(f"Error: Server responded with status code {response.status_code}.")
            print("Download failed. Check the file ID or URL.")
            return

        print(f"Saving to {destination}...")
        try:
            with open(destination, "wb") as f:
                for chunk in response.iter_content(CHUNK_SIZE):
                    if chunk: # filter out keep-alive new chunks
                        f.write(chunk)
            print("Download complete.")
        except IOError as e:
            print(f"Error: Could not write to destination file {destination}.")
            print(f"Details: {e}")


    session = requests.Session()
    params = {}

    if server == 'google':
        URL = "https://docs.google.com/uc?export=download"
        file_id = IDS.get(subset)
        if not file_id:
            print(f"Error: Google Drive ID not found for subset '{subset}'.")
            return
        params = {'id': file_id}
    elif server == 'snu':
        # Construct the URL for the SNU server
        URL = 'http://data.cv.snu.ac.kr:8008/webdav/dataset/REDS/' + subset + '.zip'
        params = {}

    print(f"Requesting URL: {URL}...")
    response = session.get(URL, params=params, stream=True)

    # Google Drive specific token check for large files
    token = get_confirm_token(response)
    if token:
        print("Google Drive warning bypass required...")
        params['confirm'] = token
        response = session.get(URL, params=params, stream=True)

    save_response_content(response, destination)


def main():
    parser = argparse.ArgumentParser(description='Download REDS dataset from google drive to current folder', allow_abbrev=False)

    parser.add_argument('--server', type=str, default='snu', choices=('google', 'snu'), help='Download server choice (default: snu).')
    parser.add_argument('--all', action='store_true', help='Download full REDS dataset (all subsets).')

    # Dynamically add arguments for all subsets from the IDS dictionary
    for subset_name in IDS.keys():
        # Using '-' in the command line flag for cleaner usage (e.g., --train-sharp)
        parser.add_argument(f'--{subset_name.replace("_", "-")}',
                            action='store_true',
                            help=f'Download {subset_name}.zip')

    # FIX for Colab/Jupyter Error:
    # If the script is run in an interactive notebook (which passes the '-f' flag),
    # we explicitly pass an empty argument list to use defaults and avoid the error.
    if 'ipykernel' in sys.modules:
        print("Detected interactive environment (Jupyter/Colab). Using default arguments to prevent '-f' error.")
        args = parser.parse_args([])
    else:
        args = parser.parse_args()

    # Determine the target directory
    if os.path.basename(os.getcwd()) == 'REDS':
        root_dir = '.'
    else:
        os.makedirs('REDS', exist_ok=True)
        root_dir = 'REDS'

    download_requested = False

    for subset in IDS:
        argdict = args.__dict__
        # Argument names are converted to snake_case by argparse, e.g., 'train-sharp' becomes 'train_sharp'
        if args.all or argdict.get(subset): # The dictionary keys match the snake_case names
            filename = '{}/{}.zip'.format(root_dir, subset)
            servername = 'Google Drive' if args.server == 'google' else 'SNU CVLab'
            print('='*50)
            print('Downloading {}.zip from {}'.format(subset, servername))
            download_file_from_server(args.server, subset, filename)
            download_requested = True

    if not download_requested:
        print('='*50)
        print("No subsets selected for download. Use '--all' or specific flags (e.g., '--train-sharp').")
        print("To see all options, run with '-h'.")

if __name__ == '__main__':
    main()
```

```python
import glob
import os


def regroup_reds_dataset(train_path, val_path):
    """Regroup original REDS datasets.

    We merge train and validation data into one folder, and separate the
    validation clips in reds_dataset.py.
    There are 240 training clips (starting from 0 to 239),
    so we name the validation clip index starting from 240 to 269 (total 30
    validation clips).

    Args:
        train_path (str): Path to the train folder.
        val_path (str): Path to the validation folder.
    """
    # move the validation data to the train folder
    val_folders = glob.glob(os.path.join(val_path, '*'))
    for folder in val_folders:
        new_folder_idx = int(folder.split('/')[-1]) + 240
        os.system(f'cp -r {folder} {os.path.join(train_path, str(new_folder_idx))}')


if __name__ == '__main__':
    # train_sharp
    train_path = 'datasets/REDS/train_sharp'
    val_path = 'datasets/REDS/val_sharp'
    regroup_reds_dataset(train_path, val_path)

    # train_sharp_bicubic
    train_path = 'datasets/REDS/train_sharp_bicubic/X4'
    val_path = 'datasets/REDS/val_sharp_bicubic/X4'
    regroup_reds_dataset(train_path, val_path)
```

```python
%%bash
cd /content/IART
cat << 'EOF' > create_video.py
import os
import torch
import cv2 # Import OpenCV
from torchvision import transforms
from PIL import Image
import numpy as np # Import numpy

def load_frames_as_numpy(folder_path):
    # This function will now return a list of numpy arrays (frames)
    frames_np = []
    # Sort files to ensure correct frame order
    image_files = sorted([f for f in os.listdir(folder_path) if f.endswith(('.png', '.jpg', '.jpeg', '.webp'))])

    if not image_files:
        print(f"No image files found in {folder_path}")
        return [], 0, 0

    # Get the dimensions from the first image to set up VideoWriter
    first_img_path = os.path.join(folder_path, image_files[0])
    first_img = Image.open(first_img_path).convert('RGB')
    width, height = first_img.size

    for filename in image_files:
        img_path = os.path.join(folder_path, filename)
        try:
            img_pil = Image.open(img_path).convert('RGB') # Ensure RGB format
            img_np = np.array(img_pil) # Convert PIL image to NumPy array (H, W, C), RGB
            frames_np.append(cv2.cvtColor(img_np, cv2.COLOR_RGB2BGR)) # Convert RGB to BGR for OpenCV
        except Exception as e:
            print(f"Error loading image {img_path}: {e}")
            continue

    return frames_np, width, height

def main():
    input_frames_path = '/content/IART/demo/Vid4_BI_results_2/foliage'
    output_video_dir = '/content/IART/demo/Vid4_BI_results_2/vid'
    output_video_path = os.path.join(output_video_dir, 'foliage_sr.mp4')

    os.makedirs(output_video_dir, exist_ok=True)

    print(f"Loading frames from: {input_frames_path}")
    frames, width, height = load_frames_as_numpy(input_frames_path)

    if not frames:
        print("No frames found or loaded. Cannot create video.")
        return

    print(f"Creating video: {output_video_path}")
    fps = 30 # Default fps for Vid4 is usually 25 or 30. Let's assume 30 for now.
    # Define the codec and create VideoWriter object
    fourcc = cv2.VideoWriter_fourcc(*'mp4v') # Codec for .mp4 files
    out = cv2.VideoWriter(output_video_path, fourcc, fps, (width, height))

    if not out.isOpened():
        print(f"Error: Could not open video writer for path {output_video_path}")
        return

    for frame in frames:
        out.write(frame)

    out.release() # Release the video writer
    print("Video created successfully.")

if __name__ == '__main__':
    main()
EOF
source activate /usr/local/envs/IART
python create_video.py
```

```python
%%bash
cd /content/IART
cat << 'EOF' > calculate_psnr.py
import os
import numpy as np
from PIL import Image
from skimage.metrics import peak_signal_noise_ratio

def load_frames(folder_path):
    frames = []
    image_files = sorted([f for f in os.listdir(folder_path) if f.endswith(('.png', '.jpg', '.jpeg', '.bmp'))])
    for filename in image_files:
        img_path = os.path.join(folder_path, filename)
        try:
            img = Image.open(img_path).convert('RGB')
            frames.append(np.array(img)) # Load as NumPy array
        except Exception as e:
            print(f"Error loading image {img_path}: {e}")
    return frames

def main():
    original_low_res_frames_path = '/root/.cache/kagglehub/datasets/uom200647r/vid4-dataset/versions/1/BIx4/calendar'
    super_resolved_frames_path = '/content/drive/MyDrive/Vid4_BI_results2/calendar'

    print(f"Loading original low-resolution frames from: {original_low_res_frames_path}")
    original_low_res_frames = load_frames(original_low_res_frames_path)

    print(f"Loading super-resolved frames from: {super_resolved_frames_path}")
    super_resolved_frames = load_frames(super_resolved_frames_path)

    if not original_low_res_frames or not super_resolved_frames:
        print("Could not load enough frames to calculate PSNR.")
        return

    # Debugging: Print dimensions of the first frame from each list
    if original_low_res_frames:
        print(f"Dimensions of first original low-res frame: {original_low_res_frames[0].shape}")
    if super_resolved_frames:
        print(f"Dimensions of first super-resolved frame: {super_resolved_frames[0].shape}")

    # Manually scale down super-resolved frames
    scaled_down_sr_frames = []
    target_height, target_width, _ = original_low_res_frames[0].shape
    print(f"Scaling down super-resolved frames to target dimensions: ({target_height}, {target_width})")

    for sr_frame_np in super_resolved_frames:
        sr_frame_pil = Image.fromarray(sr_frame_np)
        # Resize using LANCZOS for high-quality downsampling
        scaled_down_pil = sr_frame_pil.resize((target_width, target_height), Image.LANCZOS)
        scaled_down_sr_frames.append(np.array(scaled_down_pil))

    if len(original_low_res_frames) != len(scaled_down_sr_frames):
        print(f"Warning: Mismatch in number of frames after scaling. Original LR: {len(original_low_res_frames)}, Scaled SR: {len(scaled_down_sr_frames)}")
        min_frames = min(len(original_low_res_frames), len(scaled_down_sr_frames))
        original_low_res_frames = original_low_res_frames[:min_frames]
        scaled_down_sr_frames = scaled_down_sr_frames[:min_frames]

    psnr_values = []
    for i in range(len(original_low_res_frames)):
        try:
            # PSNR expects values in [0, 255] for uint8 images
            psnr = peak_signal_noise_ratio(original_low_res_frames[i], scaled_down_sr_frames[i], data_range=255)
            psnr_values.append(psnr)
        except Exception as e:
            print(f"Error calculating PSNR for frame {i}: {e}")

    if psnr_values:
        avg_psnr = np.mean(psnr_values)
        print(f"\nAverage PSNR for 'calendar' video (Original LR vs. Scaled-down SR): {avg_psnr:.2f} dB")
    else:
        print("No PSNR values could be calculated.")

if __name__ == '__main__':
    main()
EOF
source activate /usr/local/envs/IART
python calculate_psnr.py
```
